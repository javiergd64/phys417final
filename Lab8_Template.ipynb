## Understanding the Data

a) Data size and Structure: The data consists of 32,298 labeled images split into 28,709 training images and 3,589 testing images. There are 7 different classes an image could belong to and the training and test consist of about even amounts of each class except for "disgust" which has far fewer samples in both the training and test datasets.  
b) Data cleanliness: All of the images have already been resized and cropped so that the face fills about the same amount of space in each image. However, the angles for each face are not necessarily the same, some of the images have watermarks on them, and some images appear to have a lower resolution than others. The images were labeled by human volunteers but there isn't a guarentee that the people in the images were actually feeling those emotions, just that a human volunteer perceived it as such.   
c) Each image is a 48x48 black and white image. So all the images consist of the same number of features. As mentioned in a), most of the labels have the same number of datapoints but disgust has far fewer.

## Developing an initial plan for your project

a) The inputs of the model will be the images themselves in their 48x48 form. The output for each sample will be the 1D length 7 vector of the per-class probabilities for that sample. For data preparation we plan to rescale the pixel values from being between 0 and 255 to between 0 and 1 as well as applying standard scaling. To make our model more rigorous we also plan on rotating or flipping some of the images (this also increases our training data).
b) Since we're working with image data, we're going to use a Convolutional Neural Network (although we also plan on having some fully connected layers at the end)
c) Since we're doing a classification task we're going to use Cross-Entropy Loss.
d) We plan on evaluating our model on the test data and also we'll have a real-time component and we'll be able to make faces and see if the classification makes sense.

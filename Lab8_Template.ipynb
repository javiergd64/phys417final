## Understanding the Data

a) Data size and Structure: The data consists of 32,298 labeled images split into 28,709 training images and 3,589 testing images. There are 7 different classes an image could belong to and the training and test consist of about even amounts of each class except for "disgust" which has far fewer samples in both the training and test datasets.  
b) Data cleanliness: All of the images have already been resized and cropped so that the face fills about the same amount of space in each image. However, the angles for each face are not necessarily the same, some of the images have watermarks on them, and some images appear to have a lower resolution than others
